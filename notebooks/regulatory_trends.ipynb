```python
import pandas as pd
reg = pd.read_csv('../data/perma_registry/perma_registry.csv')
print('Total PERMA:', len(reg))
print(reg['current_legal_status'].value_counts())
```

```python
print('PERMA by Year:')
print(reg['year'].value_counts().sort_index())
print('\nDocument Types:')
print(reg['document_type'].value_counts())
```
